## ASCII code to String Conversion

```python
text = "Hey Vidhi! Have a good day ahead."
```

```python
lis=[]
for char in text:
    lis.append(ord(char)) 
```

```python
lis
```

```python
#to get the text back from the ASCII representation
```

```python
text1 = ""
for i in lis:
    text1 += chr(i)
```
